## Geometry: Square

```python
square_entropy_eta_2 = pd.read_csv("data/entropy_values_square_100_2.csv")
square_entropy_eta_3 = pd.read_csv("data/entropy_values_square_100_3.csv")
square_entropy_eta_4 = pd.read_csv("data/entropy_values_square_100_4.csv")
```

```python
plt.figure(figsize=(6,4))
plt.plot(np.array(square_entropy_eta_2['r0']), np.array(square_entropy_eta_2['entropy']), label=r'$\eta=2$', color='C0')
plt.plot(np.array(square_entropy_eta_3['r0']), np.array(square_entropy_eta_3['entropy']), label=r'$\eta=3$', color='C2')
plt.plot(np.array(square_entropy_eta_4['r0']), np.array(square_entropy_eta_4['entropy']), label=r'$\eta=4$', color='C1')
plt.xlabel(r'$r_0$')
plt.ylabel(r'$H(\mathcal{G})$')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(np.array(square_entropy_eta_2['r0']), np.array(square_entropy_eta_2['entropy']), label=r'$\eta=2$', color='C0')
plt.plot(np.array(square_entropy_eta_3['r0']), np.array(square_entropy_eta_3['entropy'])*3/2, label=r'$\eta=3$', color='C2')
plt.plot(np.array(square_entropy_eta_4['r0']), np.array(square_entropy_eta_4['entropy'])*2, label=r'$\eta=4$', color='C1')
plt.xlabel(r'$r_0$')
plt.ylabel(r'$H(\mathcal{G})$')
plt.legend()
plt.show()
```

## Geometry: Wedge of different angles

```python
wedge_eta_2_theta_pi_2 = pd.read_csv("data/entropy_values_wedge_1.570000_100_2.csv")
wedge_eta_2_theta_pi_4 = pd.read_csv("data/entropy_values_wedge_0.785000_100_2.csv")
wedge_eta_2_theta_pi_8 = pd.read_csv("data/entropy_values_wedge_0.392600_100_2.csv")
```

```python
plt.figure(figsize=(6,4))
plt.rcParams.update({'font.size': 15})
plt.plot(np.array(wedge_eta_2_theta_pi_2['r0']), np.array(wedge_eta_2_theta_pi_2['entropy']), label=r'$\theta = \frac{\pi}{2}$', color='C2')
plt.plot(np.array(wedge_eta_2_theta_pi_4['r0']), np.array(wedge_eta_2_theta_pi_4['entropy']), label=r'$\theta = \frac{\pi}{4}$', color='C0')
plt.plot(np.array(wedge_eta_2_theta_pi_8['r0']), np.array(wedge_eta_2_theta_pi_8['entropy']), label=r'$\theta = \frac{\pi}{8}$', color='C1')
plt.xlabel(r'$r_0$')
plt.ylabel(r'$H(\mathcal{G})$')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(np.array(wedge_eta_2_theta_pi_2['r0']), np.array(wedge_eta_2_theta_pi_2['entropy']), label=r'$\theta = \frac{\pi}{2}$', color='C2')
plt.plot(np.array(wedge_eta_2_theta_pi_4['r0'])*np.sqrt(2), np.array(wedge_eta_2_theta_pi_4['entropy']), label=r'$\theta = \frac{\pi}{4}$', color='C0')
plt.plot(np.array(wedge_eta_2_theta_pi_8['r0'])*2, np.array(wedge_eta_2_theta_pi_8['entropy']), label=r'$\theta = \frac{\pi}{8}$', color='C1')

plt.xticks(fontsize=15)
plt.yticks(fontsize=15)

plt.xlabel(r'$r_0\sqrt{2\theta/\pi}$', fontsize=15)
plt.ylabel(r'$H(\mathcal{G})$', fontsize=15)
plt.legend(fontsize=15)
plt.show()
```

### Entropy mass

```python
wedge_mass_0_05_2 = pd.read_csv("data/mass_values_wedge_0.783000_0.050000_2.csv")
wedge_mass_0_5_2 = pd.read_csv("data/mass_values_wedge_0.783000_0.500000_2.csv")
wedge_mass_1_2 = pd.read_csv("data/mass_values_wedge_0.783000_1.000000_2.csv")
```

```python
wedge_mass_0_05_2['x'] = np.array(wedge_mass_0_05_2['r']) * np.cos(np.array(wedge_mass_0_05_2['theta']))
wedge_mass_0_05_2['y'] = np.array(wedge_mass_0_05_2['r']) * np.sin(np.array(wedge_mass_0_05_2['theta']))

wedge_mass_0_5_2['x'] = np.array(wedge_mass_0_5_2['r']) * np.cos(np.array(wedge_mass_0_5_2['theta']))
wedge_mass_0_5_2['y'] = np.array(wedge_mass_0_5_2['r']) * np.sin(np.array(wedge_mass_0_5_2['theta']))

wedge_mass_1_2['x'] = np.array(wedge_mass_1_2['r']) * np.cos(np.array(wedge_mass_1_2['theta']))
wedge_mass_1_2['y'] = np.array(wedge_mass_1_2['r']) * np.sin(np.array(wedge_mass_1_2['theta']))
```

```python
fig, ax = plt.subplots(1,3,figsize=(15,4))
ax[0].scatter(wedge_mass_0_05_2['x'], wedge_mass_0_05_2['y'], c=wedge_mass_0_05_2['mass'], s=1, cmap='viridis')
ax[1].scatter(wedge_mass_0_5_2['x'], wedge_mass_0_5_2['y'], c=wedge_mass_0_5_2['mass'],s=1, cmap='viridis')
ax[2].scatter(wedge_mass_1_2['x'], wedge_mass_1_2['y'], c=wedge_mass_1_2['mass'],s=1, cmap='viridis')
ax[0].set_title(r"$r_0=0.05$", fontsize=18)
ax[1].set_title(r"$r_0=0.5$", fontsize=18)
ax[2].set_title(r"$r_0=1$", fontsize=18)
ax[0].axis('off')
ax[1].axis('off')
ax[2].axis('off')
ax[0].axis('equal')
ax[1].axis('equal')
ax[2].axis('equal')

cb0 = fig.colorbar(ax[0].collections[0], ax=ax[0], shrink=0.8)
cb1 = fig.colorbar(ax[1].collections[0], ax=ax[1], shrink=0.8)
cb2 = fig.colorbar(ax[2].collections[0], ax=ax[2], shrink=0.8)

cb0.ax.tick_params(labelsize=18)
cb1.ax.tick_params(labelsize=18)
cb2.ax.tick_params(labelsize=18)

plt.tight_layout()

#ax[0].colorbar()
plt.show()
```

```python
cantor_entropy_6_1_4_4 = pd.read_csv("data/entropy_values_cantor_6_0.250000_100_4.csv")
cantor_entropy_6_1_3_4 = pd.read_csv("data/entropy_values_cantor_6_0.333000_100_4.csv")
cantor_entropy_6_1_6_4 = pd.read_csv("data/entropy_values_cantor_6_0.166000_100_4.csv")
cantor_entropy_6_1_10_4 = pd.read_csv("data/entropy_values_cantor_6_0.100000_100_4.csv")
```

```python
# finer plots
cantor_entropy_1_10 = pd.read_csv("data/entropy_values_cantor_7_0.100000_100_4.csv")
cantor_entropy_1_4 = pd.read_csv("data/entropy_values_cantor_6_0.250000_100_4.csv")
cantor_entropy_1_6 = pd.read_csv("data/entropy_values_cantor_7_0.166000_100_4.csv")
cantor_entropy_1_3 = pd.read_csv("data/entropy_values_cantor_7_0.333000_100_4.csv")
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,4.5))
ax[0].plot(np.array(cantor_entropy_6_1_3_4['r0']), np.array(cantor_entropy_6_1_3_4['entropy']), label=r'$\alpha=3$', color='C0')
ax[0].plot(np.array(cantor_entropy_6_1_4_4['r0']), np.array(cantor_entropy_6_1_4_4['entropy']), label=r'$\alpha=4$', color='C1')
ax[0].plot(np.array(cantor_entropy_6_1_6_4['r0']), np.array(cantor_entropy_6_1_6_4['entropy']), label=r'$\alpha= 6$', color='C2')
ax[0].plot(np.array(cantor_entropy_6_1_10_4['r0']), np.array(cantor_entropy_6_1_10_4['entropy']), label=r'$\alpha=10$', color='C3')
ax[0].set_xlabel(r'$r_0$')
ax[0].set_ylabel(r'$H(\mathcal{G})$')

ax[1].loglog(cantor_entropy_1_3['r0'], cantor_entropy_1_3['entropy'], label=r'$\alpha= 3$', color='C0')
ax[1].loglog(cantor_entropy_1_4['r0'], cantor_entropy_1_4['entropy'], label=r'$\alpha= 4$', color='C1')
ax[1].loglog(cantor_entropy_1_6['r0'], cantor_entropy_1_6['entropy'], label=r'$\alpha= 6$', color='C2')
ax[1].loglog(cantor_entropy_1_10['r0'], cantor_entropy_1_10['entropy'], label=r'$\alpha= 10$', color='C3')
ax[1].set_xlabel(r'$r_0$', fontsize=18)
ax[1].set_ylabel(r'$H(\mathcal{G})$', fontsize=18)
ax[1].set_xlim(0.001,1)
ax[1].set_ylim(0.00001,1)
ax[1].legend()
# ax[1].set_yscale('log')
# ax[1].set_xscale('log')

plt.tight_layout()
ax[0].legend()
plt.show()
```
